```python
import numpy as np
def norm(X,gamma,beta,nums=1e-5):
    mean=np.mean(X,axis=-1,keepdims=True)
    varience=np.var(X,axis=-1,keepdims=True)
    X_norm=((X-mean)/np.sqrt(varience+nums))
    return X_norm*gamma+beta

X=np.random.randn(7,4)
gamma=np.ones(4)
beta=np.zeros(4)
out=norm(X,gamma,beta)
print(out.shape)
```

```python
def softmax(x):
    exp=(x-np.max(x,axis=-1,keepdims=True))
    return (exp/np.sum(exp,axis=-1,keepdims=True))

def multihead(X,num_head):
    model_dim=X.shape[1]
    head=model_dim//num_head
    heads=[]
    for i in range(num_head):
        wq=np.random.randn(model_dim,head)
        wv=np.random.randn(model_dim,head)
        wk=np.random.randn(model_dim,head)

        Q=X@wq
        K=X@wk
        V=X@wv
        attention=Q@K.T
        score=softmax(attention/np.sqrt(head))
        out=score@V
        heads.append(out)
    concatenate=np.concatenate(heads,axis=-1)
    wo=np.random.randn(model_dim,model_dim)
    final=concatenate@wo
    return final

X=np.random.randn(7,4)
mul=multihead(X,num_head=2)
print(mul.shape)
```

```python
def relu(x):
     return np.maximum(0,x)
    

def ffn(X,hidden=16):
    d=X.shape[1]
    w1=np.random.randn(d,hidden)
    b1=np.zeros(hidden)
    w2=np.random.randn(hidden,d)
    b2=np.zeros(d)
    return relu(X@w1+b1)@w2 + b2
```

```python
gamma=np.ones(4)
beta=np.zeros(4)
x=X+pe
for _ in range(6):
    x=norm(x+multihead(x,2),gamma,beta)
    x=norm(x+ffn(x),gamma,beta)

print(x.shape)
```
